```python
import numpy as np
marks=np.array([65,72,81,55,90])
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
print("Std.Dev:",marks.std())
```

```python
import pandas as pd

data={
    "Name":["A","B","C","D","E"],"Marks":[65,72,81,55,90]
}

df=pd.DataFrame(data)
print(df)
print(df["Marks"].mean())
```

```python
import matplotlib.pyplot as plt
plt.bar(df["Name"],df["Marks"])
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student marks")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.hist(df["Marks"])
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of marks")
plt.show()
```

```python
import numpy as np
n=1000
result=np.random.choice([1,2,3,4,5,6],size=n)
event1= np.sum(result>3)
event2=np.sum(result%2==0)
probgrt3=event1/n
probeven=event2/n
print("The Probability of getting a number greater than 3 is ",probgrt3)
print("The Probability of getting an even number is ",probeven)
eventand = np.sum((result > 3) & (result % 2 == 0))
proband=eventand/n
condprob=proband/probgrt3
print("The Conditional Probability of getting an even number knowing the number is greater than 3 is",condprob)
```
