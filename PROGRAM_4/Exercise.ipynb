```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
def r_squared(y, y_hat):
    ss_res = np.sum((y - y_hat)**2)
    ss_tot = np.sum((y - np.mean(y))**2)
    return 1 - ss_res/ss_tot
```

QUESTION 1

```python
# --- Data ---
x = np.linspace(1, 10, 20)
y = 2.5*x + 1.8 + np.random.normal(0, 5, 20)

# --- 1. LINEAR FIT: Normal Equations ---
n = len(x)
a = (n*np.sum(x*y) - np.sum(x)*np.sum(y)) / (n*np.sum(x**2) - np.sum(x)**2)
b = (np.sum(y) - a*np.sum(x)) / n
print(f'Linear: y = {a:.4f}x + {b:.4f}')
y_lin = a*x + b
print(f'R² = {r_squared(y, y_lin):.4f}')

# --- VISUALIZATION (Single Model) ---
fig, ax = plt.subplots(figsize=(8, 6))

xf = np.linspace(x.min(), x.max(), 200)
ax.scatter(x, y, color='blue', s=50, label='Data')
ax.plot(xf, a*xf + b, 'r-', lw=2, label=f'y={a:.2f}x+{b:.2f}')

for xi, yi, yp in zip(x, y, y_lin):
    ax.plot([xi, xi], [yi, yp], 'g--', alpha=0.4)

ax.set_title(f'Linear Fit R²={r_squared(y, y_lin):.4f}')
ax.legend()

plt.tight_layout()
plt.show()
```

```python
# --- 3. EXPONENTIAL FIT: Linearize ln(y) = ln(a) + bx ---
x3 = np.linspace(0, 3, 20)
y3 = 2.0*np.exp(-0.8*x3) + np.random.normal(0, 0.02, 20)
log_y3  = np.log(y3)
ce = np.polyfit(x3, log_y3, 1)
a_e, b_e = np.exp(ce[1]), ce[0]
print(f'Exponential: y = {a_e:.4f}•e^({b_e:.4f}x)')
y_exp = a_e * np.exp(b_e * x3)
print(f'R² = {r_squared(y3, y_exp):.4f}')

# --- VISUALIZATION (Single Model) ---
fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(x3, y3, color='blue', s=50, label='Data')
ax.plot(x3, y_exp, 'r-', lw=2, label='Exponential Fit')

ax.set_title(f'Exponential Fit  R²={r_squared(y3, y_exp):.4f}')
ax.legend()
plt.tight_layout(); plt.show()
```

```python
# --- 4. POWER FIT: Linearize ln(y) = ln(a) + b•ln(x) ---
x4 = np.linspace(0.5, 5, 20)
y4 = 1.5*x4**1 + np.random.normal(0, 0.5, 20)
cp = np.polyfit(np.log(x4), np.log(y4), 1)
a_p, b_p = np.exp(cp[1]), cp[0]
print(f'Power: y = {a_p:.4f}•x^{b_p:.4f}')
y_pow = a_p * x4**b_p
print(f'R² = {r_squared(y4, y_pow):.4f}')

# --- VISUALIZATION (Single Model) ---

fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(x4, y4, color='blue', s=50, label='Data')
ax.plot(x4, y_pow, 'r-', lw=2, label='Power Fit')

ax.set_title(f'Power Fit  R²={r_squared(y4, y_pow):.4f}')
ax.legend()
plt.tight_layout(); plt.show()
```
